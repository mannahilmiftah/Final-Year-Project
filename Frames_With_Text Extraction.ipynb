```python
import pandas as pd
import numpy as np
import random
import math
from PIL import Image
import torch
import cv2
import json
import os
from tqdm import tqdm
```

```python
VIDEO_DIR = r'C:\Users\hp\Desktop\7thSemester\FYP\datasets\ego4D\videos\v2\full_scale'
NARRATIONS_FILE = r'C:\Users\hp\Desktop\7thSemester\FYP\datasets\ego4D\videos\v2\full_scale\narrations.json'
SAVE_DIR = r'C:\Users\hp\Desktop\7thSemester\FYP\datasets\ego4D\Images\\'
```

```python
narrations = dict()
f = open(NARRATIONS_FILE)
narrations = json.load(f)
videoIds = os.listdir(VIDEO_DIR)
videoIds = videoIds[:-3]
```

```python
framesWithText = {}

for video_id in videoIds:
    video = video_id[:-4]
    if video in narrations:
        frames_with_text = {}
        for narration in narrations[video]:
            frame_number = narration['frame']
            text = narration['text']
            frames_with_text.update({frame_number: text})
        framesWithText[video] = frames_with_text
    else:
        print(f"Narrations not found for videoId: {video_id}")
```

```python
for videoId in videoIds:  
    videoName = videoId[:-4]
    video_path = os.path.join(VIDEO_DIR, videoId)
    video = cv2.VideoCapture(video_path)
    fps = int(video.get(cv2.CAP_PROP_FPS))
    totalFrames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))
    for frameNo in tqdm(range(totalFrames)):
        ret, frame = video.read()
        if ret:
            if frameNo in framesWithText[videoName].keys():
                img = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))   
                img.save(os.path.join(SAVE_DIR, f'{videoName}_{frameNo}.jpg'))
                
    video.release()
```
